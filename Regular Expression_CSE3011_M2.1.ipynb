```python
# Instead of checking text character by character, regex allows us to search patterns easily.

#Common Regex Functions
re.search()	Searches first occurrence
re.match()	Checks from beginning of string
re.findall()	Returns all matches
re.finditer()	Returns iterator of matches
re.sub()	Replaces matched text
re.split()	Splits string using regex
```

```python
# re.search(), Searches the entire string.

import re

text = "I love Python programming."
result = re.search("Python", text)
if result:
    print("Match Found")
```

```python
import re

txt = "The rain in Spain"
x = re.search("Portugal", txt)
print(x)
```

```python
# re.match() Matches only at the beginning.

import re

text = "Python Programming"
print(re.match("Python", text))
```

```python
# re.findall(), Returns all matches.

import re

text = "cat bat mat cat"
result = re.findall("cat", text)
print(result)
```

```python
# re.sub(), Replace text.

import re

text = "I like Java"
new_text = re.sub("Java", "Python", text)

print(new_text)
```

```python
import re
text = "cat bat rat"

for match in re.finditer(r"\w+at", text):
    print(match.group(), match.start())
```

```python
# re.split(), Split using regex.

import re

text = "Apple,Orange;Banana Mango"
print(re.split("[,; ]", text))
```

```python
import re

txt = "The rain in Spain"
x = re.split("\s", txt)
print(x)
```

```python
# Important Regex Symbols
```

```python
# Dot (.) Matches any single character.

import re

text = "cat bat hat"
print(re.findall(".at", text))
```

```python
# Caret (^) Beginning of string.

import re

text = "Python Programming"
print(bool(re.search("^Python", text)))
```

```python
# Dollar ($) End of string.

import re

text = "Learn Python"
print(bool(re.search("Python$", text)))
```

```python
# Star (*) Zero or more repetitions.

import re

text = "ac abc abbc abbbc"
print(re.findall("ab*c", text))
```

```python
# Plus (+), One or more repetitions.

import re

text = "ac abc abbc"
print(re.findall("ab+c", text))
```

```python
# Question Mark (?) Zero or one repetition.

import re

text = "color colour"
print(re.findall("colou?r", text))
```

```python
# Character Classes
```

```python
import re

text = "Roll Numbers: 101 102 103"
print(re.findall("\d+", text))   #\d : digit
print(re.findall("[A-Za-z]+", text))   #[A-Za-z]+ : alphabets
print(re.findall("[a-z]+", text))   #[a-z]+ : lower case
print(re.findall("[A-Z]+", text))   #[A-Z]+ : Upper case

print(re.findall("\D+", text))   #\D : Not digit
print(re.findall("\d{2,4}", text))   #\d{2, 4} : 2-4 digit

```

```python
Q1: Validate Mobile Number (10 digits)
```

```python
import re

mobile = "9876543210"

if re.fullmatch("[0-9]{10}", mobile):
    print("Valid")
else:
    print("Invalid")
```

```python
Q2: Validate Email
```

```python
import re

email = "student@gmail.com"

pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

if re.fullmatch(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
```
